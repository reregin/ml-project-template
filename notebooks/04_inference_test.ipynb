This notebook exercises the authoritative shared prediction implementation. It does not recreate preprocessing or model-loading logic. Run it after installing the package and training the example model.

```python
import json
import time
from pathlib import Path

from ml_project_template.config import get_settings
from ml_project_template.predict import predict_record

settings = get_settings()
```

## Predict one raw record

The sample contains an unseen categorical value to verify that the fitted encoder handles new categories.

```python
request_path = Path('../examples/binary_classification/sample_request.json')
record = json.loads(request_path.read_text(encoding='utf-8'))
record
```

```python
started_at = time.perf_counter()
result = predict_record(record, settings.model_artifact_path)
latency_ms = (time.perf_counter() - started_at) * 1000

print(result)
print(f'Latency: {latency_ms:.2f} ms')
```
